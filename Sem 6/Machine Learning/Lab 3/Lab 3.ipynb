## Step 1. Import the necessary libraries

```python
import pandas as pd
```

## Step 2. Import the dataset

```python
df = pd.read_csv('50_Startups.csv')
df.head()
```

## Step 3 . Check the State Column

```python
df['State'].info()
```

## Step 4 . Splitting dataset in to input and output

```python
input = df[['R&D Spend', 'Administration', 'Marketing Spend', 'State']]
output = df['Profit']
```

## Step 5 . Convert state Column into Numeric Column

## Step 5.1 .  Import  ColumnTransformer

```python
from sklearn.compose import ColumnTransformer
```

## Step 5.2 .  Import  OneHotEncoder

```python
from sklearn.preprocessing import OneHotEncoder
```

## Step 5.3 .  Perform Transformation

```python
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])])
```

```python
input_transformed = ct.fit_transform(input)
input_transformed
```

## Step 6 . Dummy variable trap

```python
input = input[:, 1:]
print(input)
```

## Step 7 Splitting dataset in to Train and Test

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(input_transformed, output, test_size=0.2, random_state=0)
```

## Step 8  Import LinearRegression model from linear_model family 

```python
from sklearn.linear_model import LinearRegression
```

## Step 9  Fit the data

```python
lr = LinearRegression()

lr.fit(X_train, y_train)
```

## Step 10  Predict the data

```python
pred = lr.predict(X_test)
```

## Step 11  Display Result

```python
print(pred)
```

```python
import matplotlib.pyplot as plt

plt.plot(y_test, pred, 'o')
```
