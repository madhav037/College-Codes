## Step 1. Import the necessary libraries

```python
import pandas as pd
```

## Step 2. Import the dataset

```python
iphone_df = pd.read_csv('./iphones.csv')
iphone_df
```

## Step 3. Import LinearRegression model from linear_model family 

```python
from sklearn.linear_model import LinearRegression
```

## Step 4. Create LinearRegression Model

```python
lr = LinearRegression()
```

## Step 5. Splitting dataset in to input and output

```python
x, y = iphone_df['iphonenumber'].values.reshape(-1, 1), iphone_df['price']
```

## Step 6. Fit the Model (Training)

```python
lr.fit(x, y)
```

## Step 7. Prdict the value

```python
# predict_y = lr.predict([[14]])
predict_Y = lr.predict(x)
```

## Step 8. Result Visualization

```python
import matplotlib.pyplot as plt
plt.plot(iphone_df['iphonenumber'], iphone_df['price'],c='r', marker='*')
plt.plot(iphone_df['iphonenumber'], predict_Y, c='b', marker='+')
```

## Dataset - 2 - Salary Dataset

## Step 1. Import the necessary libraries

```python
salary_df = pd.read_csv('./Salary_Data.csv')
from sklearn.model_selection import train_test_split
```

## Step 2.  Read the Data

```python
salary_df
```

## Step 3. View Data using Scatter plot 

```python
plt.scatter(salary_df['YearsExperience'], salary_df['Salary'])
```

## Step 4. Splitting dataset in to input and output

```python
X_train, x_test, Y_train, y_test = train_test_split(salary_df[['YearsExperience']], salary_df[['Salary']], test_size=1/3, random_state=0)
```

## Step 4. Splitting dataset in to input and output

```python
X_train, Y_train
```

```python
x_test, y_test
```

## Step 6. Create LinearRegression Model

```python
lrm = LinearRegression()
```

## Step 7. Fit the Model (Training)

```python
lrm.fit(X_train, Y_train)
```

##  Step 6. Predicting the Test set results

```python
predict = lrm.predict(x_test)
```

##  Step 7. Visualising the Training set results

```python
plt.scatter(X_train,Y_train,c='red')
plt.plot(X_train, lrm.predict(X_train),marker='*')
plt.title("Salary vs Experience (Training set)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
```

##  Step 8. Visualising the Test set results

```python
plt.scatter(x_test,y_test,c='red')
plt.plot(X_train, lrm.predict(X_train),marker='*')
plt.title("Salary vs Experience (Training set)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
```
