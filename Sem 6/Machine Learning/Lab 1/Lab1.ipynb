### Step 2. Replace missing values with mean and median

```python
gender_cat = titanic.groupby('Sex')
gender_mean = gender_cat.Age.mean()
for d, data in gender_cat:
    if d == 'male':
        titanic.loc[titanic['Sex'] == 'male', 'Age'] = titanic.loc[titanic['Sex'] == 'male', 'Age'].fillna(gender_mean['male'])
    else:
        titanic.loc[titanic['Sex'] == 'female', 'Age'] = titanic.loc[titanic['Sex'] == 'female', 'Age'].fillna(gender_mean['female'])

titanic['Age'].head(10)
```
