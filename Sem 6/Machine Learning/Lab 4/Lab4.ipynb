# Importing the libraries

```python
import pandas as pd
import numpy as np
```

# Importing the libraries

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

# Plot the data using scatter plot

```python
df = pd.read_csv('Position_Salaries.csv')
df.head()
```

```python
plt.scatter(df['Level'], df['Salary'])
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(df[['Level']], df['Salary'], test_size=0.2, random_state=42)
```

# Fitting Linear Regression to the dataset

```python
lr = LinearRegression()

lr.fit(X_train, y_train)
```

# Predict the x_test using Linear Model

```python
pred = lr.predict(X_test)
```

# Visualising the Linear Regression results

```python
plt.scatter(X_train, y_train)   
plt.plot(X_test, pred, color='red')
```

# Fitting Polynomial Regression to the dataset

## import  PolynomialFeatures

```python
from sklearn.preprocessing import PolynomialFeatures
```

## create PolynomialFeatures of degree 4

```python
poly = PolynomialFeatures(degree=4)
```

## Apply fit_transform to features

```python
X_poly = poly.fit_transform(df[['Level']])
```

## View new  Features

```python
plt.scatter(df['Level'], df['Salary'])
```

## Fitting Polynomial Regression to the dataset

```python
lr.fit(X_poly, df['Salary'])
```

# Create Grid for higher resolution and smoother curve

```python
import numpy as np

X_grid = np.arange(min(df['Level']), max(df['Level']), 0.1)
X_grid = X_grid.reshape((len(X_grid), 1))

X_grid_poly = poly.transform(X_grid)

y_grid_pred = lr.predict(X_grid_poly)

# Plot the results
plt.scatter(df['Level'], df['Salary'], color='red')
plt.plot(X_grid, y_grid_pred, color='blue')
plt.title('Polynomial Regression')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()
```
